```python
%load_ext autoreload
%autoreload 2

from skimage import transform

from fundus_stitching_toolkit.io import read_imgs_from_folder
from fundus_stitching_toolkit.registration.process import Descriptors, FundusRegistration, Keypoints, ReferenceChoice
from fundus_stitching_toolkit.utils.visu import imshow
```

```python
examples = 2
data = read_imgs_from_folder(f"../../data/examples/{examples}")


# First parameter: list of images
# Second parameter: reference choice. If ReferenceChoice.OD, the reference image is the one with the OD closest to the center of the image
# If ReferenceChoice.MACULA, the reference image is the one with the macula closest to the center of the image
# If ReferenceChoice.MAX_VESSELS, the reference image is the one with the most vessels

# Third parameter: descriptor choice. If Descriptors.TOPO_BASED, the descriptor is the vector from the center of the OD to vessels crossing.
# If Descriptors.ORB, the descriptor is based on ORBs.
f = FundusRegistration(
    data,
    ReferenceChoice.MAX_VESSELS,
    Descriptors.ORB,
    keypoints_type=Keypoints.VESSELS_CROSSING,
    transform=transform.SimilarityTransform,
    max_ratio=0.9,
    alpha_stitch=50,
)
```

```python
# Just for visualization purposes
f.extract_keypoints()
# The blue box indicates the reference image.
f.plot(with_masks=True, with_keypoints=True)
```

```python
# Left is the reference image

f.save_result(f"../../data/output/examples_{examples}/img.png", with_metadata=True)
```

```python
# Blending is done based on the distant transform of the warped ROI, defined as d
# It is then smoothed by passing it to a sigmoid function with alpha as a parameter (sig(d*alpha))


f.alpha_stitch = 5
r = f.stitch()
_ = imshow(r)
```
